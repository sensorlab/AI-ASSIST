```
import joblib
import pandas as pd
```

```
lf = pd.read_pickle("../data/interscada/pl/interim/lf.pkl")
```

```
state_cols = joblib.load("../data/interscada/pl/interim/topology_cols.joblib.z")
```

```
lf[list(state_cols)].drop_duplicates(keep="last")
```

```
import json

import requests

# Extract first sample
sample = lf.iloc[0]

state_dict = state_dict = {k: (None if pd.isna(v) else v) for k, v in sample.to_dict().items()}

res = requests.post(
    "http://localhost:8000/api/v1/estimate/tsa/by-location",
    json={
        "variant": "1.0.0",
        "state": state_dict,
        "exclude_uids": [str(sample.name)],
    },
)

res.raise_for_status()
```

```
with open("sample_interscada_pl__by_location.json", mode="w") as fp:
    json.dump(res.json(), fp, indent=2)
```
