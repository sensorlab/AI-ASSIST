```
from pathlib import Path
from pprint import pprint

import joblib
import numpy as np
import pandas as pd
from sklearn.metrics import mean_absolute_error, mean_squared_error
```

```
PROJECT_ROOT = Path().resolve().parent
REPORT_PATH = PROJECT_ROOT / "report-ml-regression-2026-05-29.joblib"

assert REPORT_PATH.exists()
```

```
data = joblib.load(REPORT_PATH)
len(data)
```

```
df = pd.concat([data["summary"], _data["summary"]])
```

```
columns = [pair for pair in df.columns if pair[-1] == "mean"]
df[columns]
```

```
columns = [pair for pair in df.columns if pair[-1] == "mean"]
df = df.loc[:, columns]

df.columns = [c[0] for c in df.columns]

df
```

```
mapper: dict[str, str] = {
    "extra_trees": "ExtraTrees",
    "random_forest": "RandomForest",
    "hist_gradient_boosting": "HistGradientBoost",
    "location_median": "LocationMedian",
    "global_median": "GlobalMedian",
    "service_location_then_global": "AI-ASSIST",
}

df.index = df.index.map(mapper)
```
